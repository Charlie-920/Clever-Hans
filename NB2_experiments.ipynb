# NB2 — White Patch Experiment

## Research question

Does a CNN trained on Fashion-MNIST images with class-specific white patches rely on patch position when classifying images?

## Test conditions

We will use the same 10,000 test images and their original labels in three conditions:

1. Clean images with no patch.
2. Images with a patch in the correct class position.
3. Images with a patch in an incorrect class position.

## Hypotheses

**H1:** Accuracy will be higher with correctly placed patches than with no patches.

**H2:** Incorrectly placed patches will reduce accuracy compared with correctly placed patches. Some predictions may follow the class indicated by the incorrect patch.

For an image of class y, the incorrect patch will use the position for class (y + 1) % 10.
